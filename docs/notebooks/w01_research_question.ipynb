## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load dataset
df = pd.read_csv("/content/content_refresh_anonymized.csv")

# Key metrics
total_pages = len(df)
declining_pages = len(df[df["trend_direction"] == "down"])
declining_pct = (declining_pages / total_pages) * 100
high_impact_pages = len(
    df[(df["trend_direction"] == "down") & (df["impressions_90d"] >= 1000)]
)

print(f"Total articles: {total_pages:,}")
print(f"Declining articles: {declining_pages:,} ({declining_pct:.1f}%)")
print(f"High-impact targets (>=1,000 impressions): {high_impact_pages:,}")
```

Analyzing the 30,000 mature articles in the starter dataset shows that 16,262 pages (54.2%) are currently in an active downward traffic trend. Out of those declining pages, 8,031 are high-impact assets that generated at least 1,000 impressions over the last 90 days. Because an editorial team cannot manually inspect over 8,000 decaying pages, an automated priority ranking system is essential.
